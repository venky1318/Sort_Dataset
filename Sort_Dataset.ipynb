```python
# Task 20: Sort Dataset Using Pandas

## Objective
Sort numerical and categorical columns in ascending and
descending order and demonstrate sorting by multiple columns.

## Deliverables
- Ascending sorting example
- Descending sorting example
- Categorical column sorting
- Multiple-column sorting
- Explanation of each sorting operation
```

```python
import pandas as pd
import numpy as np

data = {
    "student_id": [
        1001, 1002, 1003, 1004, 1005, 1006,
        1007, 1008, 1009, 1010, 1011, 1012
    ],
    "name": [
        "Aarav", "Diya", "Kabir", "Ananya",
        "Ishaan", "Meera", "Rohan", "Sara",
        "Vivaan", "Aditi", "Arjun", "Nisha"
    ],
    "department": [
        "Computer Science", "Electronics", "Mechanical",
        "Computer Science", "Civil", "Electronics",
        "Mechanical", "Computer Science", "Civil",
        "Electronics", "Mechanical", "Computer Science"
    ],
    "age": [20, 21, 20, 22, 21, 20, 22, 21, 20, 22, 21, 20],
    "marks": [86, 91, 74, 95, 68, 88, 79, 93, 72, 84, 77, 90],
    "attendance": [92, 88, 79, 96, 72, 90, 85, 94, 80, 86, 75, 91]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df)
```

```python
print("STUDENTS SORTED BY MARKS — ASCENDING")

ascending_marks = df.sort_values(
    by="marks",
    ascending=True
)

display(ascending_marks)
```

```python
print("STUDENTS SORTED BY MARKS — DESCENDING")

descending_marks = df.sort_values(
    by="marks",
    ascending=False
)

display(descending_marks)
```

```python
print("DEPARTMENTS — ASCENDING ALPHABETICAL ORDER")

department_ascending = df.sort_values(
    by="department",
    ascending=True,
    kind="stable"
)

display(department_ascending)
```

```python
print("DEPARTMENTS — DESCENDING ALPHABETICAL ORDER")

department_descending = df.sort_values(
    by="department",
    ascending=False,
    kind="stable"
)

display(department_descending)
```

```python
print("MULTIPLE-COLUMN SORTING")
print("Department: Ascending | Marks: Descending")

multi_column_sort = df.sort_values(
    by=["department", "marks"],
    ascending=[True, False]
)

display(multi_column_sort)
```

```python
print("MARKS DESCENDING, THEN ATTENDANCE DESCENDING")

marks_attendance_sort = df.sort_values(
    by=["marks", "attendance"],
    ascending=[False, False]
)

display(marks_attendance_sort)
```

```python
sorting_summary = pd.DataFrame({
    "Operation": [
        "Marks ascending",
        "Marks descending",
        "Department ascending",
        "Department descending",
        "Department ascending + marks descending"
    ],
    "Sorting column(s)": [
        "marks",
        "marks",
        "department",
        "department",
        "department, marks"
    ],
    "Ascending setting": [
        "True",
        "False",
        "True",
        "False",
        "[True, False]"
    ],
    "Number of records": [
        len(ascending_marks),
        len(descending_marks),
        len(department_ascending),
        len(department_descending),
        len(multi_column_sort)
    ]
})

display(sorting_summary)
```
